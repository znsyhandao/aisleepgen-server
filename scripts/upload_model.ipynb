```python
from google.colab import drive
import shutil
import os

# 挂载Google Drive
drive.mount('/content/drive')

# 创建models文件夹 (如果不存在)
models_dir = '/content/drive/MyDrive/models'
os.makedirs(models_dir, exist_ok=True)

# 从本地上传（需先压缩）
# 运行此单元格后会显示上传按钮
from google.colab import files
uploaded = files.upload()

# 将上传的文件移动到models文件夹
for filename in uploaded.keys():
    destination = os.path.join(models_dir, filename)
    shutil.move(filename, destination)
    print(f"已移动 {filename} 到 {destination}")

# 或者从URL直接下载到Google Drive
model_url = "你的模型下载URL"
!wget -O "/content/drive/MyDrive/models/Janus-1.3B.tar.gz" "{model_url}"

# 解压到指定目录
!tar -xzvf "/content/drive/MyDrive/models/Janus-1.3B.tar.gz" -C "/content/drive/MyDrive/models/"
```
